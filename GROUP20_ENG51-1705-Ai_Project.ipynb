```python
import os
from ultralytics import YOLO

# 1. กำหนด Path ชี้ตรงไปยังภาพ
base_path = os.path.abspath("my_dataset").replace("\\", "/")
train_images = f"{base_path}/train/images"

yaml_content = f"""
path: {base_path}
train: {train_images}
val: {train_images}

names:
  0: motorcycle
"""

with open("parking_data.yaml", "w", encoding="utf-8") as f:
    f.write(yaml_content.strip())

# 2. เริ่ม Train โดยสั่งรันบน RTX 3080 (device=0)
model = YOLO("yolo11n.pt")
results = model.train(
    data="parking_data.yaml",
    epochs=300,
    imgsz=640,
    batch=16,
    device=0,
    patience=100,       # เพิ่มบรรทัดนี้: ถ้า 50 รอบหลังความแม่นยำไม่ดีขึ้น จะหยุดเทรนทันทีเพื่อกัน Overfitting
    name="sut_model_gpu"
)

print("🎉 เทรนด้วย GPU เสร็จเรียบร้อย!")
```
